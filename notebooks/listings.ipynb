# Airbnb NYC 2019 — Data Exploration
This notebook explores the raw Airbnb dataset to understand its structure, distributions, and key patterns.

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.dataloader import load_data

df = load_data()
print('Shape:', df.shape)
```

## 2. Missing Values

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

## 3. Price Distribution

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['price'].clip(0, 500), bins=60, kde=True, color='steelblue')
plt.title('Price Distribution (clipped at $500)')
plt.xlabel('Price ($)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## 4. Listings by Room Type

```python
plt.figure(figsize=(8, 4))
df['room_type'].value_counts().plot(kind='bar', color='coral', edgecolor='black')
plt.title('Listings by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 5. Price by Neighbourhood Group

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df[df['price'] < 500], x='neighbourhood_group', y='price', palette='Set2')
plt.title('Price by Neighbourhood Group (price < $500)')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Price ($)')
plt.tight_layout()
plt.show()
```

## 6. Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## 7. Top 10 Most Listed Neighbourhoods

```python
plt.figure(figsize=(10, 5))
df['neighbourhood'].value_counts().head(10).plot(kind='barh', color='mediumpurple')
plt.title('Top 10 Neighbourhoods by Listing Count')
plt.xlabel('Number of Listings')
plt.tight_layout()
plt.show()
```
